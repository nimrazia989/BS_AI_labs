```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
temperature = [30, 32, 31, 35, 33]

plt.plot(days, temperature, marker='o')
plt.title("Temperature Changes Over 5 Days")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["Ali", "Sara", "Ahmed", "Ayesha"]
marks = [78, 85, 90, 72]

colors = ["blue", "green", "orange", "purple"]

plt.bar(students, marks, color=colors)

plt.title("Students Marks Comparison")

for i in range(len(students)):
    plt.text(i, marks[i] + 1, str(marks[i]), ha='center')

plt.show()
```

# TASK NO 7

```python


import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

names = ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"]
marks = [78, 85, 90, 72, 88]
attendance = [80, 92, 95, 75, 89]

data = pd.DataFrame({
    "Names": names,
    "Marks": marks,
    "Attendance": attendance
})

print(data)

# Bar Chart for Marks
plt.bar(names, marks)

plt.title("Students Marks")

plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()

# Line Chart for Attendance
plt.plot(names, attendance, marker='o')

plt.title("Attendance Trend")

plt.xlabel("Students")
plt.ylabel("Attendance %")

plt.grid(True)

plt.show()

# Heatmap
heatmap_data = data[["Marks", "Attendance"]]

sns.heatmap(heatmap_data.corr(), annot=True)

plt.title("Marks vs Attendance Relationship")

plt.show()
```
