# Data Cleaning

```python
print("Customers Missing Values")
print(customers.isnull().sum())

print("Purchases Missing Values")
print(purchases.isnull().sum())

print("Membership Missing Values")
print(membership.isnull().sum())
```

# Fill Missing Values

```python
customers["Age"] = customers["Age"].fillna(
    customers["Age"].median()
)

customers["Income"] = customers["Income"].fillna(
    customers["Income"].mean()
)

customers["City"] = customers["City"].fillna(
    "Unknown"
)

purchases["Amount"] = purchases["Amount"].fillna(
    purchases["Amount"].median()
)
```

# Duplicate Removal

```python
customer_duplicates = customers.duplicated().sum()
purchase_duplicates = purchases.duplicated().sum()

customers.drop_duplicates(inplace=True)
purchases.drop_duplicates(inplace=True)

print("Customer Duplicates Removed:", customer_duplicates)
print("Purchase Duplicates Removed:", purchase_duplicates)
```

# Standardize City Names

```python
customers["City"] = customers["City"].str.strip().str.title()

customers["Gender"] = customers["Gender"].str.upper()

membership["MembershipType"] = (
    membership["MembershipType"]
    .str.strip()
    .str.title()
)
```

# Outlier Handling 

```python
customers.loc[customers["Age"] < 18, "Age"] = 18
customers.loc[customers["Age"] > 70, "Age"] = 70 
```

```python
customers.loc[customers["Income"] < 0, "Income"] = customers["Income"].median()
```

```python
purchases.loc[purchases["Quantity"] <= 0, "Quantity"] = 1
```

# Dataset Merging
Merge Customers + Purchases

```python
merged1 = pd.merge(
    customers,
    purchases,
    on="CustomerID",
    how="left"
)
```

# Merge with Membership

```python
final_df = pd.merge(
    merged1,
    membership,
    on="CustomerID",
    how="left"
)
```

 # Merge Statistics 

```python
print("Shape:", final_df.shape)

print("Matched Records:",
      final_df["MembershipType"].notnull().sum())

print("Unmatched Records:",
      final_df["MembershipType"].isnull().sum())

print("Null Values:")
print(final_df.isnull().sum())
```

 # Feature Engineering
Age Group 

```python
def age_group(age):
    if age <= 30:
        return "Young"
    elif age <= 50:
        return "Adult"
    else:
        return "Senior"

final_df["AgeGroup"] = final_df["Age"].apply(age_group)
```

# Income Category 

```python
def income_category(income):
    if income < 40000:
        return "Low"
    elif income <= 80000:
        return "Medium"
    else:
        return "High"

final_df["IncomeCategory"] = (
    final_df["Income"].apply(income_category)
)
```

# Spending Score 

```python
final_df["SpendingScore"] = (
    final_df["Amount"] *
    final_df["Quantity"]
)
```

# Customer Value 

```python
def customer_value(score):
    if score > 100000:
        return "High Value"
    elif score >= 50000:
        return "Medium Value"
    else:
        return "Low Value"

final_df["CustomerValue"] = (
    final_df["SpendingScore"]
    .apply(customer_value)
)
```

# Purchase Frequency 

```python
frequency = purchases.groupby(
    "CustomerID"
).size()

final_df["PurchaseFrequency"] = (
    final_df["CustomerID"]
    .map(frequency)
)
```

# Membership Duration 

```python
membership["JoinDate"] = pd.to_datetime(
    membership["JoinDate"]
)

today = pd.Timestamp.today()

membership["MembershipDuration"] = (
    today -
    membership["JoinDate"]
).dt.days

final_df = pd.merge(
    final_df,
    membership[["CustomerID",
                "MembershipDuration"]],
    on="CustomerID",
    how="left"
)
```

 # Machine Learning Preparation
Remaining Missing Values 

```python
print(final_df.isnull().sum())
```

# Encode Categorical Variables 

```python
encoder = LabelEncoder()

categorical_cols = [
    "Gender",
    "City",
    "Product",
    "MembershipType",
    "AgeGroup",
    "IncomeCategory",
    "CustomerValue"
]

for col in categorical_cols:
    final_df[col] = encoder.fit_transform(
        final_df[col].astype(str)
    ) 
```

# Normalize Numerical Features 

```python
scaler = MinMaxScaler()

final_df[
    ["Income",
     "Amount",
     "SpendingScore"]
] = scaler.fit_transform(
    final_df[
        ["Income",
         "Amount",
         "SpendingScore"]
    ]
) 
```

# Remove Irrelevant Columns 

```python
final_df.drop(
    ["Name", "PurchaseID"],
    axis=1,
    inplace=True,
    errors="ignore"
)
```
