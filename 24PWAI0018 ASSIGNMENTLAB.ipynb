```python
import numpy as np

data = np.array([
    [78, 85, 80, 90, 1],
    [45, 50, 40, 55, 0],
    [88, 92, 84, 91, 1],
    [60, 58, 65, 62, 1],
    [35, 40, 38, 45, 0],
    [95, 90, 93, 97, 1],
    [72, 75, 70, 68, 1],
    [50, 48, 52, 47, 0],
    [81, 79, 85, 88, 1],
    [42, 45, 40, 43, 0]
])

X = data[:, :-1]
y = data[:, -1]

print("Features (X):")
print(X)

print("\nLabels (y):")
print(y)

average_marks = np.mean(X)
maximum_marks = np.max(X)
minimum_marks = np.min(X)

print("\nAverage Marks:", average_marks)
print("Maximum Marks:", maximum_marks)
print("Minimum Marks:", minimum_marks)

passed_students = np.sum(y == 1)
failed_students = np.sum(y == 0)

print("\nNumber of Passed Students:", passed_students)
print("Number of Failed Students:", failed_students)

above_80 = X[X > 80]

print("\nMarks Above 80:")
print(above_80)

below_50 = X[X < 50]

print("\nMarks Below 50:")
print(below_50)

X_normalized = (X - np.min(X)) / (np.max(X) - np.min(X))

print("\nNormalized Dataset:")
print(X_normalized)

average_per_student = np.mean(X, axis=1)

predictions = np.where(average_per_student >= 60, 1, 0)

print("\nPredicted Results:")
print(predictions)

accuracy = np.mean(predictions == y) * 100

print("\nPrediction Accuracy:", accuracy, "%")
```
